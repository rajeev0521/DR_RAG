# 05 - Ablation Arm: Fixed-Size Chunks Baseline
Executes controlled evaluation across random seeds using frozen configuration.

```python
from stp_rag.experiments.pipeline import run_ablation_arm, compute_file_sha256

# 1. Assert Dev Configuration Integrity
CONFIG_PATH = "configs/stp_params.yaml"
current_sha = compute_file_sha256(CONFIG_PATH)
print(f"Active config SHA-256: {current_sha}")
```

```python
# 2. Execute Ablation Arm across 3 Random Seeds
seeds = [42, 123, 999]
results = []

for s in seeds:
    print(f"Running Fixed-Size Chunks Baseline with seed {s}...")
    res = run_ablation_arm(
        variant="fixed_size",
        seed=s,
        split="test",
        use_in_memory_store=True,
    )
    results.append(res)
    print(f"  Seed {s} -> Recall@5: {res.recall_at_5:.3f}, MRR: {res.mrr:.3f}, EM: {res.em:.3f}, F1: {res.f1:.3f}")
```

```python
# 3. View Saved Results
from stp_rag.eval.results_schema import load_all
df = load_all("results/")
arm_df = df[df["system"] == "fixed_size"]
print(arm_df[["system", "seed", "recall_at_5", "mrr", "em", "f1", "chunks_per_doc"]])
```
